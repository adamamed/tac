## Imports

```python
import json
import requests
```

```python
titre = "Bruxelles"
url = f"https://fr.wikipedia.org/api/rest_v1/page/summary/{titre}"
headers = {"User-Agent": "TAC-TP1 (etudiant ULB)"}
resp = requests.get(url, headers=headers)
print(resp.status_code)
data = resp.json()
print(data["title"])
print(data["extract"])
```

```python
headers = {"User-Agent": "TAC-TP1 (etudiant ULB)"}

def resume_wikipedia(titre):
    url = f"https://fr.wikipedia.org/api/rest_v1/page/summary/{titre}"
    resp = requests.get(url, headers=headers)
    if resp.status_code != 200:
        return None
    data = resp.json()
    return {"titre": data["title"], "resume": data["extract"]}

lieux = ["Bruxelles", "Liège", "Anvers"]
for lieu in lieux:
    resultat = resume_wikipedia(lieu)
    if resultat:
        print(resultat["titre"])
        print(resultat["resume"][:200])
        print("-" * 40)
```
