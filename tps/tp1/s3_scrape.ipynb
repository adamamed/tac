# Scraping : téléchargement des PDF CAMille

Ce notebook récupère la liste des 51 PDF du corpus CAMille sur le site de l'ULB et les télécharge dans le dossier data/pdf.

## Imports

```python
import os
import re
import time
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
from urllib.parse import urljoin

url = "https://max.de.wilde.web.ulb.be/camille/"
page = requests.get(url)
soup = BeautifulSoup(page.text, "html.parser")
liens = [urljoin(url, a["href"]) for a in soup.find_all("a", href=True)
         if a["href"].lower().endswith(".pdf")]
print(len(liens))
```

```python
dossier = "../../data/pdf"
os.makedirs(dossier, exist_ok=True)
for lien in liens:
    nom = lien.split("/")[-1]
    r = requests.get(lien)
    with open(os.path.join(dossier, nom), "wb") as f:
        f.write(r.content)
    time.sleep(0.5)
print("Terminé")
```
